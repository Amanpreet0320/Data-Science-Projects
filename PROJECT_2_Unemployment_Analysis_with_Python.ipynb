## 1. Import Python Libraries

The required Python libraries are imported for data manipulation, numerical analysis, and visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
```

## 2. Load the Unemployment Datasets

Two datasets are loaded for analysis:
- **Unemployment in India** – provides regional unemployment information.
- **Unemployment Rate upto 11_2020** – provides unemployment information covering the COVID-19 period.

These datasets will be explored and compared to identify important unemployment patterns.

```python
import pandas as pd

india_data = pd.read_csv("Unemployment in India.csv")
covid_data = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("India Dataset Shape:", india_data.shape)
print("COVID-19 Period Dataset Shape:", covid_data.shape)
```

## 3. Explore the Datasets

Let's inspect the first few records of both datasets to understand their structure and the type of unemployment information available.

```python
print("India Unemployment Dataset")

display(india_data.head())

print("COVID-19 Period Unemployment Dataset")

display(covid_data.head())
```

## 4. Datasets Information

The structure, number of records, and data types of both datasets are examined to understand the available variables before performing further analysis.

```python
print("India Unemployment Dataset Information")
india_data.info()

print("\n" + "=" * 60 + "\n")

print("COVID-19 Period Unemployment Dataset Information")
covid_data.info()
```

## 5. Statistical Summary

Descriptive statistics are calculated for both datasets to examine the central tendency and overall distribution of the numerical unemployment indicators.

```python
print("Summary of India Unemployment Data")

display(india_data.describe().round(2))

print("\n" + "=" * 60 + "\n")

print("Summary of COVID-19 Period Data")

display(covid_data.describe().round(2))
```

## 6. Missing Value Analysis

Before performing further analysis, we check both datasets for missing values. This helps ensure that the visualizations and statistical results are based on reliable data.

```python
india_missing = india_data.isnull().sum()
covid_missing = covid_data.isnull().sum()

print("Missing Values - India Unemployment Dataset")
display(india_missing.to_frame(name="Missing Count"))

print("\nMissing Values - COVID-19 Unemployment Dataset")
display(covid_missing.to_frame(name="Missing Count"))
```

```python
# Identify rows containing missing values

india_missing_rows = india_data[india_data.isnull().any(axis=1)]

print("Rows with Missing Values in India Dataset:", india_missing_rows.shape)
display(india_missing_rows.head())
```

## 7. Data Cleaning

To prepare the datasets for analysis, rows containing missing values are removed. This ensures that the subsequent statistical analysis and visualizations are based on complete observations.

```python
india_data = india_data.dropna().copy()
covid_data = covid_data.dropna().copy()

print("Cleaned India Dataset Shape:", india_data.shape)
print("Cleaned COVID Dataset Shape:", covid_data.shape)
```

### Verification of Missing Values After Cleaning

The datasets are checked again to confirm that no missing values remain after data cleaning.

```python
print("Missing Values After Cleaning - India Dataset")
display(india_data.isnull().sum())

print("\nMissing Values After Cleaning - COVID Dataset")
display(covid_data.isnull().sum())
```

```python
# Standardize column names by removing extra spaces

india_data.columns = india_data.columns.str.strip()
covid_data.columns = covid_data.columns.str.strip()

print("India Dataset Columns:")
print(india_data.columns.tolist())

print("\nCOVID Dataset Columns:")
print(covid_data.columns.tolist())
```

## 8. Date Formatting

The date columns are converted into a consistent datetime format to make the datasets easier to analyze and compare over time.

```python
india_data["Date"] = pd.to_datetime(india_data["Date"], dayfirst=True)
covid_data["Date"] = pd.to_datetime(covid_data["Date"], dayfirst=True)

print("India Dataset Date Type:", india_data["Date"].dtype)
print("COVID-19 Dataset Date Type:", covid_data["Date"].dtype)
```

```python
print("India Dataset")

print("Start Date:", india_data["Date"].min())
print("End Date:", india_data["Date"].max())

print("\nCOVID-19 Dataset")

print("Start Date:", covid_data["Date"].min())
print("End Date:", covid_data["Date"].max())
```

## 9. Verify Clean Datasets

The cleaned datasets are verified by checking their structure, column information, and data types.

```python
print("Missing Values in First Dataset")

display(india_data.isnull().sum())

print("\nMissing Values in Second Dataset")

display(covid_data.isnull().sum())

print("\nRows with Missing Values:")

print(india_data[india_data.isnull().any(axis=1)].shape)
```

```python
print("First Dataset")

display(india_data.head())

print("\nSecond Dataset")

display(covid_data.head())
```

```python
print("Statistical Summary of First Dataset")

display(india_data.describe().round())

print("\nStatistical Summary of Second Dataset")

display(covid_data.describe().round())
```

## 10. Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the distribution of unemployment rates and identify important patterns and variations in the data.

### Distribution of Estimated Unemployment Rate

This histogram shows how estimated unemployment rates are distributed across the observations in the dataset. It helps identify the most common unemployment range, the overall spread of the data, and the presence of extreme values.

```python
plt.figure(figsize=(9, 5.5))

sns.histplot(
    india_data["Estimated Unemployment Rate (%)"],
    bins=18,
    kde=True,
    color="purple",
    edgecolor="black",
    linewidth=1
)

plt.xlabel("Estimated Unemployment Rate (%)", fontsize=11)
plt.ylabel("Number of Observations", fontsize=11)
plt.title(
    "Distribution of Estimated Unemployment Rate",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

### Urban vs Rural Unemployment Rate

This box plot compares unemployment rates between urban and rural areas. It highlights differences in the distribution and the presence of outliers in each area.

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data=india_data,
    x="Area",
    y="Estimated Unemployment Rate (%)",
    palette='magma'
)

plt.xlabel("Area")
plt.ylabel("Estimated Unemployment Rate (%)")
plt.title("Urban vs Rural Unemployment Rate", fontsize=14, fontweight="bold", pad=12)

plt.tight_layout()
plt.show()
```

### Average Unemployment Rate by Area

This bar chart compares the average unemployment rate in urban and rural regions. It provides a simple comparison of unemployment levels across the two areas.

```python
area_avg = india_data.groupby("Area")["Estimated Unemployment Rate (%)"].mean()

plt.figure(figsize=(7, 5))

area_avg.plot(
    kind="bar",
    color=["#4B2E83","#C06C84"],
    edgecolor="black"
)

plt.title("Average Unemployment Rate by Area", fontsize=14, fontweight="bold", pad=12)
plt.xlabel("Area")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## Average Monthly Unemployment Rate

This line chart shows the average unemployment rate over time. It helps identify monthly trends, fluctuations, and significant changes in unemployment levels during the study period.

```python
monthly_unemployment = (
    india_data.groupby("Date")["Estimated Unemployment Rate (%)"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_unemployment["Date"],
    monthly_unemployment["Estimated Unemployment Rate (%)"],
    marker="o",
    linewidth=2
)

plt.title("Average Monthly Unemployment Rate", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Impact of COVID-19 on Unemployment Rate

This visualization highlights the effect of the COVID-19 lockdown on unemployment by marking the lockdown period on the monthly unemployment trend. It helps demonstrate how unemployment changed before and after the lockdown.

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_unemployment["Date"],
    monthly_unemployment["Estimated Unemployment Rate (%)"],
    marker="o",
    linewidth=2,
    label="Average Unemployment Rate"
)

plt.axvline(
    pd.to_datetime("2020-03-25"),
    linestyle="--",
    linewidth=2,
    label="COVID-19 Lockdown"
)

plt.title("Impact of COVID-19 on Unemployment Rate", fontsize=16, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Average Monthly Employment

This line chart shows the average number of employed individuals each month. It helps analyze employment trends and observe how employment levels changed over time.

```python
monthly_emp = india_data.groupby(
    india_data["Date"].dt.to_period("M")
)["Estimated Employed"].mean()

monthly_emp.index = monthly_emp.index.astype(str)

plt.figure(figsize=(14, 6))

plt.plot(
    monthly_emp.index,
    monthly_emp.values,
    marker="o",
    linewidth=2
)

plt.title("Average Monthly Employment", fontsize=16, fontweight="bold")
plt.xlabel("Month")
plt.ylabel("Estimated Employed")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Average Monthly Labour Participation Rate
This visualization displays the monthly average labour participation rate, showing how workforce participation varied over time during the study period.

```python
monthly_labour = (
    india_data.groupby(
        india_data["Date"].dt.to_period("M")
    )["Estimated Labour Participation Rate (%)"]
    .mean()
)

monthly_labour.index = monthly_labour.index.astype(str)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_labour.index,
    monthly_labour.values,
    marker="o",
    linewidth=2
)

plt.title(
    "Average Monthly Labour Participation Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel("Labour Participation Rate (%)")

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 📊 Top 10 States with Highest Average Unemployment Rate

This analysis identifies the top 10 regions with the highest average estimated unemployment rate during the study period. The horizontal bar chart makes it easy to compare regions and highlights areas that experienced relatively higher unemployment.

```python
state_unemployment = (
    india_data.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

ax = state_unemployment.sort_values().plot(
    kind="barh",
    figsize=(12, 6),
    color="#C06C84"
)

plt.title(
    "Top 10 States with Highest Average Unemployment Rate",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Average Unemployment Rate (%)", fontsize=12)
plt.ylabel("State / Region", fontsize=12)

plt.grid(axis="x", linestyle="--", alpha=0.3)

for i, value in enumerate(state_unemployment.sort_values()):
    ax.text(
        value + 0.1,
        i,
        f"{value:.2f}%",
        va="center",
        fontsize=10
    )

plt.tight_layout()
plt.show()
```

## Top 10 States by Labour Participation Rate

This horizontal bar chart shows the ten states with the highest average labour participation rate. It highlights regions where a larger proportion of the population actively participates in the workforce.

```python
labour_participation = (
    india_data.groupby("Region")["Estimated Labour Participation Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

labour_participation.sort_values().plot(
    kind="barh",
    color="#4B2E83"
)

plt.title("Top 10 States by Labour Participation Rate")
plt.xlabel("Average Labour Participation Rate (%)")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

## Top 10 States by Labour Participation Rate

This horizontal bar chart displays the ten states with the highest average labour participation rate. It provides insights into regions where a larger proportion of the population actively participated in the workforce.

```python
labour = india_data.groupby("Region")["Estimated Labour Participation Rate (%)"].mean()

top10_labour = labour.sort_values(ascending=False).head(10)

plt.figure(figsize=(12, 6))

plt.barh(
    top10_labour.index,
    top10_labour.values
)

plt.title("Top 10 States by Labour Participation Rate")
plt.xlabel("Average Labour Participation Rate (%)")
plt.ylabel("State")

plt.gca().invert_yaxis()
plt.grid(axis="x", alpha=0.3)

plt.show()
```
