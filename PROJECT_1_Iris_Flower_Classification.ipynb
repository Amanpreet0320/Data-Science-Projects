# Iris Flower Classification — An Overview

The Iris Flower Classification project focuses on using machine learning to identify different species of Iris flowers from their physical measurements.

The dataset contains measurements of sepal length, sepal width, petal length, and petal width for three Iris species: Setosa, Versicolor, and Virginica.

In this project, the dataset is explored and visualized to understand the patterns between the flower measurements and their species. A machine learning classification approach is then applied to predict the species of Iris flowers based on these features.

# Tools and Technologies

- Python — Programming language used for data analysis and machine learning.
- NumPy — Used for numerical operations and array handling.
- Pandas — Used for data loading, manipulation, and analysis.
- Matplotlib — Used to create data visualizations.
- Scikit-learn — Used for accessing the Iris dataset and building the classification model.
- Google Colab — Used as the development environment for running the project.

#1 : Import Required Libraries

The following libraries provide the essential tools for data manipulation, visualization, and accessing the Iris dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
```

#2 : Accessing the Iris Dataset

The Iris dataset is accessed using Scikit-learn and contains measurements of iris flowers that will be used for classification and further analysis.

```python
iris = load_iris()
```

# 3 : Structuring the Dataset with Pandas

The Iris features are organized into a structured Pandas DataFrame, making the dataset easier to inspect, analyze, and use in the classification workflow.

```python
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df
```

# 4 : Exploratory Data Analysis(EDA)

Exploratory analysis is performed to understand the structure, characteristics, and distribution of the Iris dataset before applying machine learning techniques.

The first few records are displayed to get an initial view of the dataset and verify that the feature values have been loaded correctly.

```python
df.head()
```

The final records are examined to verify the structure and values of the dataset at the end of the DataFrame.

```python
df.tail()
```

The DataFrame information is examined to check the number of entries, data types, and presence of non-null values in each feature.

```python
df.info()
```

The dimensions of the DataFrame are checked to confirm the number of observations and features available for analysis.

```python
(df.shape[0], df.shape[1])
```

The target labels are added to the DataFrame so that each flower measurement can be associated with its corresponding species class.

```python
df = df.assign(target=iris.target)
df.head()
```

```python
df['species'] = df['target'].map({
    0: 'setosa',
    1: 'versicolor',
    2: 'virginica'
})
df.head()
```

```python
df.describe()
```

The distribution of the target species is examined to verify the number of observations available for each Iris class.

```python
df['species'].value_counts()
```

# 5 : Visualizing the Data

Graphs are used to understand the data and see how the flower measurements are related to each other.

```python
import seaborn as sns
```

```python
sns. pairplot(df, hue='species', palette='viridis')
plt.show()
```

#6 : Preparing Features and Target

The flower measurements are separated from the target values to prepare the data for classification.

```python
x = df.drop(columns=['target', 'species'])
y = df.target
```

# 7 : Splitting the Dataset into Training and Testing

The dataset is divided into training data and testing data. 80% is used for training and 20% for testing.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=63)
```

# 8 : Model Training

A Decision Tree model is trained using the training data.



```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=63)
model.fit(x_train, y_train)
```

# 9 : Predicting Flower Classes

The trained model is used to predict the flower classes in the test data.

```python
y_pred = model.predict(x_test)
y_pred
```

# 10 : Model Evalution

The model is checked using accuracy, confusion matrix, and classification report.

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(confusion_matrix)
```

#Classification Report

This report shows the model's performance for each flower class.

```python
from sklearn.metrics import classification_report
classification_report = classification_report(y_test, y_pred)
print("Classification Report:")
print(classification_report)
```
